# A. Quantitative Analysis

```python

from scipy.stats import skew, kurtosis

# Meaningful numerical variables (excluding IDs, phone, zipcode)
num_vars = ["popularity", "danceability", "instrumentalness", "loudness", "speechiness", "valence"]

# Part 1: Statistical Analysis
for col in num_vars:
    print("="*60)
    print(f"📊 Statistical Analysis of {col}")
    print("="*60)

    # Summary statistics
    desc = df[col].describe()
    print("\nSummary Statistics:")
    print(desc)

    # Skewness & Kurtosis
    skew_val = skew(df[col].dropna())
    kurt_val = kurtosis(df[col].dropna())
    print(f"\nSkewness : {skew_val:.2f}")
    print(f"Kurtosis : {kurt_val:.2f}")

    # Interpretation (shape of distribution)
    if skew_val > 1:
        print("➡ Highly right-skewed (long tail to the right).")
    elif skew_val > 0.5:
        print("➡ Moderately right-skewed.")
    elif skew_val < -1:
        print("➡ Highly left-skewed (long tail to the left).")
    elif skew_val < -0.5:
        print("➡ Moderately left-skewed.")
    else:
        print("➡ Approximately symmetric distribution.")

    if kurt_val > 3:
        print("➡ Leptokurtic (heavy tails, outliers likely).")
    elif kurt_val < -1:
        print("➡ Platykurtic (light tails, flat distribution).")
    else:
        print("➡ Mesokurtic (close to normal).")
```

# B. Visualization using Bar plots

```python
columns_to_plot = ['track_id', 'track_name', 'artist_name', 'artwork_url', 'album_name', 'track_url', 'language']

# Define a list of colors to use for the bar plots (one per subplot)
colors = ['skyblue', 'lightgreen', 'salmon', 'lightcoral', 'lightblue', 'lightseagreen', 'plum']

# Validate columns exist in df (skip invalid ones)
valid_columns = [col for col in columns_to_plot if col in df.columns]
invalid_columns = [col for col in columns_to_plot if col not in df.columns]
if invalid_columns:
    print(f"Warning: Skipping invalid columns not in df: {invalid_columns}")
if not valid_columns:
    raise ValueError("No valid columns to plot. Check your DataFrame.")

# Update columns_to_plot to only valid ones
columns_to_plot = valid_columns

# Determine the number of rows and columns for the subplot grid
num_cols = 3
num_rows = (len(columns_to_plot) + num_cols - 1) // num_cols

# Create the subplots (wider figsize for long labels)
fig, axes = plt.subplots(num_rows, num_cols, figsize=(30, 20))  # Increased width
axes = axes.flatten()  # Flatten for easier indexing

# Function to truncate long labels (for x-ticks)
def truncate_labels(labels, max_len=20):
    return [label[:max_len] + '...' if len(str(label)) > max_len else str(label) for label in labels]

# Iterate through the columns and plot each one
for i, col in enumerate(columns_to_plot):
    ax = axes[i]

    try:
        # Drop NaNs and get top 10 categories by count
        series_clean = df[col].dropna()
        if series_clean.empty:
            ax.text(0.5, 0.5, f'No valid data for {col} (all NaN/empty)',
                    ha='center', va='center', transform=ax.transAxes, fontsize=12)
            ax.set_title(f"Bar Plot of {col}")
            continue

        top_categories = series_clean.value_counts().head(10).index
        if len(top_categories) == 0:
            raise ValueError(f"No categories after cleaning for {col}")

        # Filter data (only non-NaN rows matching top categories)
        df_filtered = df[df[col].isin(top_categories)].dropna(subset=[col])

        if df_filtered.empty:
            raise ValueError(f"Filtered data empty for {col}")

        # Build title
        title = f"Top 10: {col.replace('_', ' ').title()}"
        if df[col].nunique() > 10:
            title += ' (Top 10 shown)'

        # Create the bar plot (wider bars, sorted by count, single color per subplot)
        sns.countplot(x=col, data=df_filtered, ax=ax, color=colors[i % len(colors)],
                      order=top_categories, width=0.8)  # Wider bars for better centering

        # Get the counts for labeling (reindex to match order)
        counts = df_filtered[col].value_counts().reindex(top_categories)

        # Add count labels on top of bars (safely handle patches, with dynamic offset)
        num_patches = len(ax.patches)
        for j in range(min(num_patches, len(counts))):
            p = ax.patches[j]
            count = counts.iloc[j]
            height = p.get_height()
            if height >= 5:  # Only label significant bars
                offset = max(1, height * 0.02)  # Slightly larger offset for visibility
                # Precise centering: x-position at bar center
                x_pos = p.get_x() + p.get_width() / 2.
                ax.text(x_pos, height + offset, f'{int(height)}',
                        ha='center', va='bottom', fontsize=9, fontweight='bold')

        # Customize the subplot
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.set_xlabel(col.replace('_', ' ').title())
        ax.set_ylabel("Count")

        # Handle x-ticks: Rotate more aggressively and truncate long labels
        ax.tick_params(axis='x', rotation=60, labelsize=8)  # 60 degrees for better fit
        # Truncate and re-set tick labels
        tick_labels = ax.get_xticklabels()
        truncated_labels = truncate_labels([label.get_text() for label in tick_labels])
        ax.set_xticklabels(truncated_labels, ha='right')  # Right-align for rotated labels

    except (KeyError, ValueError, IndexError) as e:
        # Catch and handle errors per column without crashing
        ax.text(0.5, 0.5, f'Error plotting {col}: {str(e)}',
                ha='center', va='center', transform=ax.transAxes, fontsize=10, color='red')
        ax.set_title(f"Error: {col}", color='red')
        print(f"Error for column '{col}': {e}")  # Log to console
        continue
    except Exception as e:
        # Catch any other unexpected errors
        print(f"Unexpected error for column '{col}': {e}")
        ax.text(0.5, 0.5, 'Unexpected error', ha='center', va='center', transform=ax.transAxes)
        continue

# Hide/remove unused subplots
for i in range(len(columns_to_plot), len(axes)):
    fig.delaxes(axes[i])

# Adjust layout (increased spacing for middle row and long labels)
plt.tight_layout(pad=2.0)
plt.subplots_adjust(left=0.08, right=0.98, top=0.92, bottom=0.15,
                    hspace=0.4, wspace=0.25)  # More space: bottom for rotated labels, wspace for horizontal

# Show the plots
plt.show()
```

# c. Visualization using Pie plots

```python
# Define the columns to plot
columns_to_plot = ['track_id', 'track_name', 'artist_name', 'artwork_url', 'album_name', 'track_url', 'language']

# Define a list of color palettes (seaborn named palettes)
color_palettes = ['Dark2', 'Accent', 'Set2', 'Paired']  # Added more for variety/cycling

# Validate columns exist in df (skip invalid ones)
valid_columns = [col for col in columns_to_plot if col in df.columns]
invalid_columns = [col for col in columns_to_plot if col not in df.columns]
if invalid_columns:
    print(f"Warning: Skipping invalid columns not in df: {invalid_columns}")
if not valid_columns:
    raise ValueError("No valid columns to plot. Check your DataFrame.")

# Update columns_to_plot to only valid ones
columns_to_plot = valid_columns

# Determine the number of rows and columns for the subplot grid
num_cols = 4
num_rows = (len(columns_to_plot) + num_cols - 1) // num_cols

# Create the subplots (adjusted figsize for pies)
fig, axes = plt.subplots(num_rows, num_cols, figsize=(30, 20))  # Wider for 4 cols
if num_rows * num_cols == 1:
    axes = [axes]
axes = axes.flatten()

# Function to truncate long labels for pie slices
def truncate_labels(labels, max_len=15):
    return [str(label)[:max_len] + '...' if len(str(label)) > max_len else str(label) for label in labels]

# Iterate through the columns and plot each one in a separate subplot
for i, col in enumerate(columns_to_plot):
    ax = axes[i]
    ax.clear()  # Clear for safety

    try:
        # Coerce to string for safety
        df[col] = df[col].astype(str)

        # Drop NaNs and get value counts (limit to top 10 for high-cardinality)
        series_clean = df[col].dropna()
        if series_clean.empty:
            ax.text(0.5, 0.5, f'No valid data for {col}', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f"Pie Chart of {col}")
            continue

        counts_full = series_clean.value_counts()
        n_top = min(10, len(counts_full))  # Top 10 or all if fewer
        counts = counts_full.head(n_top)

        if len(counts) == 0:
            raise ValueError(f"No categories after cleaning for {col}")

        # Truncate labels
        labels = truncate_labels(counts.index)

        # Cycle through palettes
        palette = color_palettes[i % len(color_palettes)]
        colors_list = sns.color_palette(palette, n_colors=len(counts))  # Generate enough colors

        # Custom autopct: Only show % for slices >5%
        def autopct_func(pct):
            return f'{pct:.1f}%' if pct > 5 else ''

        # Create the pie chart (with explode for top slice, limited slices)
        explode = [0.05 if j == 0 else 0 for j in range(len(counts))]  # Slight explode for largest
        ax.pie(counts.values,  # Use .values for sizes
               labels=labels,
               autopct=autopct_func,
               startangle=140,
               colors=colors_list,
               explode=explode,
               textprops={'fontsize': 9, 'ha': 'center'})  # Slightly larger font

        # Set the title (note if limited)
        title = f"Pie Chart of {col.replace('_', ' ').title()}"
        if len(counts_full) > n_top:
            title += f' (Top {n_top} shown)'
        ax.set_title(title, fontsize=11, fontweight='bold')

    except (KeyError, ValueError, IndexError) as e:
        ax.text(0.5, 0.5, f'Error: {str(e)}', ha='center', va='center', transform=ax.transAxes, color='red')
        ax.set_title(f"Error: {col}", color='red')
        print(f"Error for column '{col}': {e}")
        continue
    except Exception as e:
        print(f"Unexpected error for column '{col}': {e}")
        ax.text(0.5, 0.5, 'Unexpected error', ha='center', va='center', transform=ax.transAxes)
        continue

# Remove any unused subplots safely
total_axes = len(axes)
for j in range(len(columns_to_plot), total_axes):
    try:
        fig.delaxes(axes[j])
    except ValueError:
        pass  # Already removed or invalid

# Adjust the layout safely
try:
    plt.tight_layout(pad=1.5)
except ValueError:
    pass  # Skip if layout fails

# Show the plots
plt.show(block=False)
```

#A. Utility function to create the composition stacked bar

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from typing import List, Literal, Optional


def plot_categorical_composition_stacked_bar(
    df: pd.DataFrame,
    main_category_column: str,
    composition_column: str, # This replaces sub_category_column
    aggregation_kpi: str,     # NEW: KPI to aggregate (e.g., 'Units', 'Revenue', 'Profit')
    agg_type: Literal['sum', 'mean', 'median'] = 'sum', # NEW: Aggregation type for the KPI
    n: Optional[int] = 10,
    top_or_bottom: Literal['top', 'bottom'] = 'top',
    sort_by_total_kpi: bool = True, # NEW: Whether to sort main categories by their total KPI value
    title: str = None,
    x_label: str = 'Proportion (%)',
    y_label: str = None,
    legend_title: str = None,
    colormap: str = 'viridis',
    figsize: tuple = (18, 7)
):
    """
    Generates a 100% stacked bar chart showing the proportional distribution of a
    'composition_column' (e.g., Product_Category) within 'main_category_column' (e.g., Store_Name),
    based on the aggregation of a specified KPI (e.g., Units, Revenue, Profit).

    Args:
        df (pd.DataFrame): The input DataFrame.
        main_category_column (str): The column that will form the main bars (e.g., 'Store_Name').
        composition_column (str): The column whose values will form the segments within each bar
                                  (e.g., 'Product_Category', 'Store_Location').
        aggregation_kpi (str): The name of the KPI column to aggregate (e.g., 'Units', 'Revenue', 'Profit').
        agg_type (Literal['sum', 'mean', 'median']): The aggregation type for the KPI. Defaults to 'sum'.
        n (Optional[int], optional): The number of top/bottom `main_category_column` entries to display.
                                     If None, all unique entries are displayed. Defaults to 10.
        top_or_bottom (Literal['top', 'bottom'], optional): Whether to select 'top' N or 'bottom' N
                                                             `main_category_column` entries.
                                                             Only relevant if `n` is not None. Defaults to 'top'.
        sort_by_total_kpi (bool, optional): If True, main categories are sorted by their total KPI value
                                            before selecting top/bottom N. If False, they are sorted
                                            alphabetically by `main_category_column`. Defaults to True.
        title (str, optional): Custom title for the plot. If None, a default title is generated.
        x_label (str, optional): Custom label for the x-axis. Defaults to 'Proportion (%)'.
        y_label (str, optional): Custom label for the y-axis. If None, a default label is generated.
        legend_title (str, optional): Title for the legend. If None, `composition_column` is used.
        colormap (str, optional): Matplotlib colormap to use for the stacked bars. Defaults to 'viridis'.
        figsize (tuple, optional): Figure size (width, height) in inches. Defaults to (16, 8).
    """

    # --- Input Validation ---
    if main_category_column not in df.columns:
        print(f"Error: Main category column '{main_category_column}' not found in DataFrame.")
        return
    if composition_column not in df.columns:
        print(f"Error: Composition column '{composition_column}' not found in DataFrame.")
        return
    if aggregation_kpi not in df.columns:
        print(f"Error: Aggregation KPI '{aggregation_kpi}' not found in DataFrame.")
        return
    if top_or_bottom not in ['top', 'bottom']:
        print("Error: 'top_or_bottom' must be 'top' or 'bottom'.")
        return
    if agg_type not in ['sum', 'mean', 'median']:
        print(f"Error: Invalid 'agg_type'. Must be one of: 'sum', 'mean', 'median'.")
        return
    if not pd.api.types.is_numeric_dtype(df[aggregation_kpi]):
        print(f"Error: Aggregation KPI '{aggregation_kpi}' must be a numeric column.")
        return

    # --- Group and Aggregate the KPI ---
    # Group by both main and sub categories and aggregate the KPI
    grouped_data = df.groupby([main_category_column, composition_column])[aggregation_kpi].agg(agg_type)

    # Unstack the composition_column to get the format needed for stacked bar
    # This creates a table where rows are main_categories and columns are composition_categories
    # and values are the aggregated KPI.
    pivot_table_kpi = grouped_data.unstack(fill_value=0) # fill_value=0 for combinations not present

    # --- Select Top/Bottom N Main Categories based on KPI total ---
    if n is not None:
        if sort_by_total_kpi:
            # Sort by the total (or mean, median) of the KPI for each main category
            main_category_kpi_totals = pivot_table_kpi.sum(axis=1) # Sum across all composition categories for sorting
            if top_or_bottom == 'top':
                selected_main_categories = main_category_kpi_totals.nlargest(n).index
            else: # 'bottom'
                selected_main_categories = main_category_kpi_totals.nsmallest(n).index
            # Ensure the selected categories are present in the pivot table index
            pivot_table_kpi = pivot_table_kpi.loc[pivot_table_kpi.index.isin(selected_main_categories)]
        else:
            # Sort alphabetically and then select top/bottom N
            sorted_indices = sorted(pivot_table_kpi.index)
            if top_or_bottom == 'top':
                selected_main_categories = sorted_indices[:n]
            else:
                selected_main_categories = sorted_indices[-n:]
            pivot_table_kpi = pivot_table_kpi.loc[selected_main_categories]

    # Ensure the order is based on total KPI value if sorted
    if sort_by_total_kpi:
        pivot_table_kpi = pivot_table_kpi.loc[main_category_kpi_totals.reindex(pivot_table_kpi.index).sort_values(ascending=False if top_or_bottom == 'top' else True).index]


    # --- Convert Aggregated KPI Values to Proportions (Percentages) ---
    # Divide each row's KPI values by its total KPI sum to get proportions
    # Multiply by 100 to get actual percentage values for plotting
    percentage_pivot_table = pivot_table_kpi.apply(lambda r: r / r.sum(), axis=1) * 100

    # Handle cases where a main_category might have 0 total KPI (e.g., if filtered to nothing)
    # This prevents NaN in percentage calculation if r.sum() is 0.
    percentage_pivot_table.replace([np.inf, -np.inf], np.nan, inplace=True)
    percentage_pivot_table.fillna(0, inplace=True)


    # --- Plotting ---
    sns.set_style("whitegrid")
    plt.figure(figsize=figsize)

    ax = percentage_pivot_table.plot(
        kind='barh',
        stacked=True,
        cmap=colormap,
        ax=plt.gca(),
        edgecolor='black',
        width=0.8
    )

    # --- Add Percentage Labels to Bars ---
    for container in ax.containers:
        # Filter for values > 0 to avoid cluttering with 0% labels
        labels = [f'{w:.1f}%' if w > 0.0 else '' for w in container.datavalues]
        ax.bar_label(container, labels=labels, label_type='center', fontsize=9, color='black', fontweight='bold')

    # --- Set Titles and Labels ---
    if title is None:
        title_prefix = f'{top_or_bottom.title()} {n}' if n else 'Overall'
        title = (f'{title_prefix} {main_category_column.replace("_", " ").title()} '
                 f'Composition by {composition_column.replace("_", " ").title()} '
                 f'({agg_type.title()} {aggregation_kpi.replace("_", " ").title()})')
    plt.title(title, fontsize=16, fontweight='bold')

    plt.xlabel(x_label, fontsize=12)

    if y_label is None:
        y_label = main_category_column.replace("_", " ").title()
    plt.ylabel(y_label, fontsize=12)

    # --- Customize Plot Appearance ---
    plt.xticks(fontsize=10)
    plt.yticks(fontsize=10)
    plt.xlim(0, 100) # X-axis must be 0 to 100 for percentage
    plt.grid(False)
    plt.gca().spines['right'].set_visible(False)
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['left'].set_visible(False) # Hide left spine for cleaner look

    # Add legend
    if legend_title is None:
        legend_title = composition_column.replace("_", " ").title()
    plt.legend(title=legend_title, bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)

    plt.tight_layout(rect=[0, 0, 0.85, 1])
    plt.show()
```

# Define IQR outlier detection function

```python

def detect_outliers_iqr(data, column):
    """
    Detect outliers using IQR method
    Returns outlier indices, bounds, and statistics
    """
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    # Define outlier bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]

    return {
        'outliers': outliers,
        'lower_bound': lower_bound,
        'upper_bound': upper_bound,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'outlier_count': len(outliers),
        'outlier_percentage': (len(outliers) / len(data)) * 100
    }
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# Load your dataset
df = pd.read_csv('spotify_tracks.csv')

# Select numerical audio features for outlier analysis
audio_features = [
    'acousticness', 'danceability', 'duration_ms', 'energy',
    'instrumentalness', 'key', 'liveness', 'loudness', 'mode',
    'speechiness', 'tempo', 'valence', 'popularity'
]

# Function to detect outliers using IQR method
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    return outliers, lower_bound, upper_bound

# Function to detect outliers using Z-score method
def detect_outliers_zscore(series, threshold=3):
    z_scores = np.abs(stats.zscore(series.dropna()))
    outliers = series[z_scores > threshold]
    return outliers
```

# Generate summary statistics with outlier information

```python

outlier_summary = []

for feature in audio_features:
    if feature in df.columns:
        # Basic statistics
        data = pd.to_numeric(df[feature], errors='coerce').dropna()

        # IQR method outliers
        iqr_outliers, lower_bound, upper_bound = detect_outliers_iqr(data)

        # Z-score method outliers
        zscore_outliers = detect_outliers_zscore(data)

        # Summary statistics
        summary = {
            'Feature': feature,
            'Count': len(data),
            'Mean': data.mean(),
            'Median': data.median(),
            'Std': data.std(),
            'Min': data.min(),
            'Max': data.max(),
            'Q1': data.quantile(0.25),
            'Q3': data.quantile(0.75),
            'IQR_Lower_Bound': lower_bound,
            'IQR_Upper_Bound': upper_bound,
            'IQR_Outlier_Count': len(iqr_outliers),
            'IQR_Outlier_Percentage': (len(iqr_outliers) / len(data)) * 100,
            'ZScore_Outlier_Count': len(zscore_outliers),
            'ZScore_Outlier_Percentage': (len(zscore_outliers) / len(data)) * 100,
            'IQR_Outlier_Min': iqr_outliers.min() if len(iqr_outliers) > 0 else np.nan,
            'IQR_Outlier_Max': iqr_outliers.max() if len(iqr_outliers) > 0 else np.nan
        }
        outlier_summary.append(summary)
```

# 🏭 Recommendation 2: Audio Feature Portfolio Strategy

```python

print("\n🏭 RECOMMENDATION 2: AUDIO FEATURE PORTFOLIO STRATEGY")
print("=" * 60)

# Comprehensive audio feature analysis
feature_analysis = df.agg({
    'danceability': ['mean', 'std', 'min', 'max'],
    'energy': ['mean', 'std', 'min', 'max'],
    'valence': ['mean', 'std', 'min', 'max'],
    'tempo': ['mean', 'std', 'min', 'max'],
    'acousticness': ['mean', 'std', 'min', 'max']
}).round(3)

print("\n📊 AUDIO FEATURE PORTFOLIO ANALYSIS:")
print("-" * 40)
features = ['danceability', 'energy', 'valence', 'tempo', 'acousticness']
for feature in features:
    stats = feature_analysis[feature]
    print(f"\n{feature.title()}:")
    print(f"  • Mean: {stats['mean']}")
    print(f"  • Range: {stats['min']} - {stats['max']}")
    print(f"  • Variability: {stats['std']}")
```

# Correlation with popularity

```python

print(f"\n💡 STRATEGIC INSIGHTS:")
print("-" * 40)
correlation_with_popularity = df[features + ['popularity']].corr()['popularity'].drop('popularity')
top_correlated = correlation_with_popularity.idxmax()
highest_corr = correlation_with_popularity.max()

print(f"• {top_correlated}: Most correlated with popularity (r={highest_corr:.3f})")
print(f"• Feature optimization opportunities identified")
print(f"• Portfolio balance analysis reveals diversification needs")
```

# Portfolio optimization recommendations

```python

print("\n🎯 PORTFOLIO OPTIMIZATION RECOMMENDATIONS:")
print("-" * 40)

print("1. 📈 FEATURE OPTIMIZATION MATRIX:")
for feature in features:
    corr = correlation_with_popularity[feature]
    avg_value = feature_analysis[feature]['mean']

    if corr > 0.2 and avg_value > feature_analysis[feature]['mean']:
        status = "⭐ HIGH IMPACT - Maintain & Leverage"
    elif corr > 0.2:
        status = "🚀 HIGH POTENTIAL - Increase Focus"
    elif avg_value > feature_analysis[feature]['mean']:
        status = "💰 STRONG BASE - Optimize Further"
    else:
        status = "📊 DEVELOPMENT AREA - Strategic Improvement"

    print(f"   • {feature.title()}: {status} (r={corr:.3f})")

print(f"\n2. 🎵 FEATURE-SPECIFIC ACTIONS:")
print(f"   • Priority: Enhance {top_correlated} in new productions")
print(f"   • Balance: Maintain optimal feature combinations")
print(f"   • Innovation: Experiment with underutilized feature ranges")
print(f"   • Quality: Ensure consistency across musical characteristics")

print(f"\n3. 📊 IMPLEMENTATION METRICS:")
print("   • Track feature distribution analysis")
print("   • Popularity correlation monitoring")
print("   • Audience response to feature variations")
print("   • Portfolio diversification effectiveness")
```

# Language market analysis

```python

language_performance = df.groupby('language').agg({
    'popularity': ['mean', 'count', 'sum'],
    'year': 'max',
    'danceability': 'mean',
    'energy': 'mean'
}).round(2)

language_performance.columns = ['Avg_Popularity', 'Track_Count', 'Total_Popularity',
                               'Latest_Year', 'Avg_Danceability', 'Avg_Energy']

print(f"\n📊 LANGUAGE MARKET PERFORMANCE:")
print("-" * 40)
for language in language_performance.index:
    row = language_performance.loc[language]
    print(f"\n{language}:")
    print(f"  • Avg Popularity: {row['Avg_Popularity']}/100")
    print(f"  • Total Tracks: {row['Track_Count']}")
    print(f"  • Latest Release: {int(row['Latest_Year'])}")
    print(f"  • Market Presence: {row['Track_Count']/len(df)*100:.1f}%")

# Market opportunity analysis
print(f"\n🎯 MARKET OPPORTUNITY ANALYSIS:")
print("-" * 40)
emerging_markets = language_performance[language_performance['Track_Count'] < language_performance['Track_Count'].median()]
established_markets = language_performance[language_performance['Track_Count'] >= language_performance['Track_Count'].median()]

print(f"• Established Markets: {len(established_markets)} languages")
print(f"• Emerging Markets: {len(emerging_markets)} languages")
print(f"• Growth potential in underpenetrated language markets")
```
